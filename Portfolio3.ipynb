## Product Mix Problem

```python
model = ConcreteModel()

#declare the DV
model.r=Var(domain=NonNegativeIntegers)
model.b=Var(domain=NonNegativeIntegers)

#Decalre the Objective
model.profit=Objective(expr=2.25*model.r+2.6*model.b,sense=maximize)

#Declare the constriants
model.f1=Constraint(expr=2*model.r+model.b<=4000)
model.f2=Constraint(expr=model.r+2*model.b<=5000)

SolverFactory('glpk').solve(model)

model.display()
```

## The Dual Problem

```python
model=ConcreteModel('The Dual Problem')

#declare the DV
model.f1=Var(domain=NonNegativeIntegers)
model.f2=Var(domain=NonNegativeIntegers)
model.fS=Var(domain=NonNegativeIntegers)

#Decalre the Objective
model.shadow=Objective(expr=4000*model.f1+5000*model.f2+8000*model.fS,sense=minimize)

#Declare the constriants
model.r=Constraint(expr=model.f1*2+model.f2+model.fS*3>=2.25)
model.b=Constraint(expr=model.f1+model.f2*2+model.fS*3>=2.6)

SolverFactory('glpk').solve(model)

model.display()
```

## Investment Portfolios

```python
model = ConcreteModel('Investment Portfolios')

# declare the DVs
model.g = Var(domain=NonNegativeReals)
model.i = Var(domain=NonNegativeReals)
model.m = Var(domain=NonNegativeReals)

# Declare the objective
model.yeild=Objective(expr=0.2*model.g+0.1*model.i+0.06*model.m, sense=maximize)

# Declare the constraints
model.risk=Constraint(expr=0.1*model.g+0.05*model.i+0.01*model.m<=0.05)
model.max=Constraint(expr=model.g+model.i+model.m==1)
model.gmin=Constraint(expr=model.g>=0.1)
model.imin=Constraint(expr=model.i>=0.1)
model.mmin=Constraint(expr=model.m>=0.2)

SolverFactory('glpk').solve(model)

model.display()
```

## Project Selection

```python
from pyomo.environ import *
import pandas as pd

# Create model
model = ConcreteModel('Project Selection')

# Read in data
data = pd.read_excel('Optimization Sheet.xlsx', sheet_name='ProjectSel1', index_col=0)

# Gather decision variables
P = data.columns[0:15]

# Declare decision variables
model.x = Var(P, domain=Binary)

# Declare objective function
model.z = Objective(
    expr=sum([model.x[p] * data.loc['Profit', p] for p in P]),
    sense=maximize
)

# Declare constraints
model.con = ConstraintList()
years = data.index[0:5]
for y in years:
    model.con.add(sum([model.x[p] * data.loc[y, p] for p in P]) <= data.loc[y, 'Max'])

# Solve
SolverFactory('glpk').solve(model)
model.display()
```

## Production Scheduling

```python
model=ConcreteModel('Production Scheduling')
#read in data
data=pd.read_excel('Optimization Sheet.xlsx',sheet_name='Prodline',index_col=0)
#locate DV
P=data.columns[0:2]
#Declare DVs
model.x=Var(P,domain=NonNegativeIntegers)

model.profit=Objective(expr=sum([model.x[p]*data.loc['Cases',p] for p in P])-
                       sum([model.x[p]*data.loc['chemicals',p] for p in P])-
                       sum([model.x[p]*data.loc['Cost',p] for p in P]), sense=maximize)
# Decalre Constraints
model.min_con = ConstraintList()
model.max_con = ConstraintList()
for p in P:
    model.min_con.add(model.x[p] >= data.loc['min', p])
    model.max_con.add(model.x[p] <= data.loc['max', p])

model.chem = Constraint(expr=sum([model.x[p]*data.loc['chemicals',p] for p in P]) <= 1000)

SolverFactory('glpk').solve(model)
model.display()
```

## Manufacturing and Distribution 

```python
model=ConcreteModel('Manufacturing and Distribution')
# read in data
data = pd.read_excel('Optimization Sheet.xlsx', sheet_name='MandD', index_col=0)
#Locate DVs
Plants=data.index[0:4]
Distributors=data.columns[0:7]
#Define DV
V = [(p,d) for p in Plants for d in Distributors]
model.x=Var(V,domain=NonNegativeReals)
# Define obj
model.obj = Objective(
    expr=quicksum(model.x[p,d] * (data.loc[p,'Cost'] + data.loc[p,d]) for p,d in V),
    sense=minimize)
#define Constraints
model.cons = ConstraintList()
#quick sum to removed python deprication warning
for p in Plants:
    model.cons.add(quicksum(model.x[p,d] for d in Distributors) <= data.loc[p,'Capacity'])
for d in Distributors:
    model.cons.add(quicksum(model.x[p,d] for p in Plants) >= data.loc['Dmin',d])
for d in Distributors:
    model.cons.add(quicksum(model.x[p,d] for p in Plants) <= data.loc['Dmax',d])
SolverFactory('glpk').solve(model)
model.display()
```

## Pet Food

```python
# Create model
model = ConcreteModel('Pet Food')

# Read materials data
materials_data = pd.read_excel('Optimization Sheet.xlsx', sheet_name='PetFood', index_col=0, skiprows=5, nrows=5)

products = ['Bird Food', 'Dog Food']
bird_food_materials = ['Seeds', 'Stones', 'Cereal']
dog_food_materials = ['Meat', 'Fishmeal', 'Cereal']

# Decision variables
model.x = Var(products, domain=NonNegativeReals)
model.y_bf = Var(bird_food_materials, domain=NonNegativeReals)
model.y_df = Var(dog_food_materials, domain=NonNegativeReals)

# Objective
revenue = model.x['Bird Food'] * 750 + model.x['Dog Food'] * 980
cost = sum([model.y_bf[m] * materials_data.loc['Cost $/ton', m] for m in bird_food_materials]) + \
       sum([model.y_df[m] * materials_data.loc['Cost $/ton', m] for m in dog_food_materials])
model.z = Objective(expr=revenue - cost, sense=maximize)

# Constraints
model.con = ConstraintList()

model.con.add(0.25 * model.x['Bird Food'] + 0.15 * model.x['Dog Food'] <= 8)
model.con.add(0.10 * model.x['Bird Food'] + 0.30 * model.x['Dog Food'] <= 8)

model.con.add(sum([model.y_bf[m] for m in bird_food_materials]) == model.x['Bird Food'])
model.con.add(sum([model.y_df[m] for m in dog_food_materials]) == model.x['Dog Food'])

# Bird Food composition
model.con.add(sum([model.y_bf[m] * materials_data.loc['Protein %', m] / 100 for m in bird_food_materials]) >= 5 / 100 * model.x['Bird Food'])
model.con.add(sum([model.y_bf[m] * materials_data.loc['Carbs %', m] / 100 for m in bird_food_materials]) >= 18 / 100 * model.x['Bird Food'])
model.con.add(sum([model.y_bf[m] * materials_data.loc['Trace Min %', m] / 100 for m in bird_food_materials]) >= 1 / 100 * model.x['Bird Food'])
model.con.add(sum([model.y_bf[m] * materials_data.loc['Abrasives %', m] / 100 for m in bird_food_materials]) >= 2 / 100 * model.x['Bird Food'])
model.con.add(model.y_bf['Seeds'] >= 10 / 100 * model.x['Bird Food'])

# Dog Food composition
model.con.add(sum([model.y_df[m] * materials_data.loc['Protein %', m] / 100 for m in dog_food_materials]) >= 11 / 100 * model.x['Dog Food'])
model.con.add(sum([model.y_df[m] * materials_data.loc['Carbs %', m] / 100 for m in dog_food_materials]) >= 15 / 100 * model.x['Dog Food'])
model.con.add(sum([model.y_df[m] * materials_data.loc['Trace Min %', m] / 100 for m in dog_food_materials]) >= 1 / 100 * model.x['Dog Food'])

SolverFactory('glpk').solve(model)
model.display()
```

## Employee Scheduling

```python
# Create model
model = ConcreteModel('Employee Scheduling')
# Read data
data = pd.read_excel('Optimization Sheet.xlsx', sheet_name='Sced',index_col=0)
#Locate DVs
E=data.columns[0:7]
# Convert E to a list for indexing
E_list = list(E)  # Create E_list from E

# Set DVs
model.x=Var(E,domain=NonNegativeIntegers)
# Set Obj
model.cost=Objective(expr=quicksum([model.x[e]*data.loc['Cost',e] for e in E]), sense=minimize)
combinations = {
    'tuesday': [E_list[0]] + E_list[3:7],         # 0 and 3-6
    'wednesday': E_list[0:2] + E_list[4:7],       # 0,1 and 4-6
    'thursday': E_list[0:3] + E_list[5:7],        # 0-2 and 5,6
    'friday': E_list[0:4] + [E_list[6]],          # 0-3 and 6
}
# Constraints
model.cons=ConstraintList()
model.cons.add((quicksum(model.x[e] for e in E[1:6])*1000) >= data.loc['Sunday','min'])
model.cons.add((quicksum(model.x[e] for e in E[2:7])*1000) >= data.loc['Monday','min'])
model.cons.add((quicksum(model.x[e] for e in combinations['tuesday'])*1000) >= data.loc['Tuesday','min'])
model.cons.add((quicksum(model.x[e] for e in combinations['wednesday'])*1000) >= data.loc['Wednesday','min'])
model.cons.add((quicksum(model.x[e] for e in combinations['thursday'])*1000) >= data.loc['Thursday','min'])
model.cons.add((quicksum(model.x[e] for e in combinations['friday'])*1000) >= data.loc['Friday','min'])
model.cons.add((quicksum(model.x[e] for e in E[0:5])*1000) >= data.loc['Saturday','min'])

SolverFactory('glpk').solve(model)
model.display()
```

## Lockbox Problem

```python
model = ConcreteModel('Lockbox')
# Read data
data = pd.read_excel('Optimization Sheet.xlsx', sheet_name='Lockbox', index_col=0)
Region = data.index[0:7]
City = data.columns[0:6]
# Decision variables
V = [(r, c) for r in Region for c in City]
model.x = Var(V, domain=Binary)
model.y = Var(City, domain=Binary)
# Loss/Costs
Loss = quicksum(model.x[r, c] * data.loc[r, c] for r, c in V)
Cost = quicksum(model.y[c] * data.loc['Cost', c] for c in City)
# Objective
model.obj = Objective(expr=Loss + Cost, sense=minimize)
# Constraints
model.cons = ConstraintList()
for r in Region:
    model.cons.add(quicksum(model.x[r, c] for c in City) == 1)
for r in Region:
    for c in City:
        model.cons.add(model.x[r, c] <= model.y[c])

SolverFactory('glpk').solve(model)
model.display()
```

## Farming

```python
model = ConcreteModel('Farming')
# Read data
data = pd.read_excel('Optimization Sheet.xlsx', sheet_name='Farming', index_col=0)
farms = ['Farm 1', 'Farm 2', 'Farm 3']
crops = ['Milo', 'Cotton', 'Wheat']
# Decision variables
model.x = Var(farms, crops, domain=NonNegativeReals)
# Objective
model.z = Objective(
    expr=quicksum(model.x[f, c] * data.loc['Profit', c] for f in farms for c in crops),
    sense=maximize)
# Constraints
model.con = ConstraintList()
for f in farms:
    model.con.add(quicksum(model.x[f, c] for c in crops) <= data.loc[f'{f} Acres', 'Max'])
for f in farms:
    model.con.add(quicksum(model.x[f, c] * data.loc[f'{f} Water', c] for c in crops) <= data.loc[f'{f} Water', 'Max'])
for c in crops:
    model.con.add(quicksum(model.x[f, c] for f in farms) <= data.loc['Harvest Cap', c])
SolverFactory('glpk').solve(model)
model.display()
```

## Inventory Management

```python
# Create model
model = ConcreteModel('Inventory Management')
# Read data
data = pd.read_excel('Optimization Sheet.xlsx', sheet_name='Inv', index_col=0)
days = ['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5', 'Day 6', 'Day 7']
# Decision variables
model.own_light = Var(domain=NonNegativeIntegers)
model.own_heavy = Var(domain=NonNegativeIntegers)
model.light_by_owned_light = Var(days, domain=NonNegativeReals)
model.light_by_owned_heavy = Var(days, domain=NonNegativeReals)
model.light_rented = Var(days, domain=NonNegativeReals)
model.heavy_by_owned_heavy = Var(days, domain=NonNegativeReals)
model.heavy_rented = Var(days, domain=NonNegativeReals)
# Objective
fixed_cost = (32 * model.own_light + 44 * model.own_heavy) * 7
variable_cost = quicksum(40 * model.light_by_owned_light[d] + 54 * (model.light_by_owned_heavy[d] + model.heavy_by_owned_heavy[d]) for d in days)
rental_cost = quicksum(175 * model.light_rented[d] + 225 * model.heavy_rented[d] for d in days)
model.z = Objective(expr=fixed_cost + variable_cost + rental_cost, sense=minimize)
# Constraints
model.con = ConstraintList()
for d in days:
    demand_light = data.loc['Demand - Lights', d]
    demand_heavy = data.loc['Demand - Heavies', d]
    # Meet light demand
    model.con.add(model.light_by_owned_light[d] + model.light_by_owned_heavy[d] + model.light_rented[d] >= demand_light)
    # Meet heavy demand
    model.con.add(model.heavy_by_owned_heavy[d] + model.heavy_rented[d] >= demand_heavy)
    # Can't use more owned light than owned
    model.con.add(model.light_by_owned_light[d] <= model.own_light)
    # Can't use more owned heavy than owned (for both light and heavy tasks)
    model.con.add(model.light_by_owned_heavy[d] + model.heavy_by_owned_heavy[d] <= model.own_heavy)
SolverFactory('glpk').solve(model)
model.display()
```
